## ⚙️ Data Preprocessing

```python
# Separate numerical and categorical features.
x = df.drop('claim_flag', axis= 1)
y = df['claim_flag']
```

## Data Leakage

The feature **CLM_AMT** represents the actual claim amount.

Since this information is only available after a claim occurs, using it would leak future information into the model.

Therefore, it was removed before training.

```python
x.drop(['clm_amt'], axis= 1, inplace= True)
```

- Apply different preprocessing strategies for each feature type.

```python
# RobustScaler was selected because several numerical features contain significant outliers.
# Unlike StandardScaler, it relies on the median and interquartile range, making it more robust.
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.impute import SimpleImputer
numerical_cols = ['yoj', 'car_age']
numerical_pipeline = Pipeline([('Imputer',SimpleImputer(strategy= 'median')),('scaler', RobustScaler()) ])
numerical_pipeline
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.impute import KNNImputer
numerical_cols_knn = ['kidsdriv', 'age', 'homekids', 'income', 'home_val', 'travtime','bluebook', 'tif', 'mvr_pts' ]
numerical_pipeline_knn = Pipeline([('KNN_Imputer',KNNImputer(n_neighbors=5)),('scaler', RobustScaler()) ])
numerical_pipeline_knn
```

```python
# One-Hot Encoding
# Used for low-cardinality categorical features.
from sklearn.preprocessing import OneHotEncoder
onehot_cols = x.select_dtypes(include='object').columns.drop(['education','occupation'])
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore',drop='first')
onehot_pipeline = Pipeline([('Imputer',SimpleImputer(strategy= 'constant', fill_value= 'Other')),('onehot', ohe) ])
onehot_pipeline
```

```python
# Ordinal Encoding
# Used for education because it has a natural order.
from sklearn.preprocessing import OrdinalEncoder
ord_cols =['education']
ord_pipeline = Pipeline([('ordinal_encoder',  OrdinalEncoder(categories= [ ['<High School', 'High School', 'Bachelors', 'Masters', 'PhD'] ])) ])
ord_pipeline
```

```python
# Binary Encoding
# Used for occupation because it contains many unique categories.
from category_encoders import BinaryEncoder
be_cols =['occupation']
be_pipeline = Pipeline([('ordinal_encoder',  BinaryEncoder()) ])
be_pipeline
```

```python
from sklearn.compose import ColumnTransformer
preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_pipeline, numerical_cols),
    ('num_knn',numerical_pipeline_knn,numerical_cols_knn),
    ('cat', onehot_pipeline, onehot_cols),
    ('ord',ord_pipeline,ord_cols),
    ('be',be_pipeline,be_cols)], remainder='passthrough', force_int_remainder_cols=False)
preprocessor
```

```python
import warnings

warnings.simplefilter("ignore", UserWarning)
```

```python
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
from sklearn.pipeline import Pipeline

# Create Stratified CV
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models = [
    ('Logistic Regression', LogisticRegression(random_state=42 ,class_weight='balanced')),
    ('Decision Tree', DecisionTreeClassifier(random_state=42 ,class_weight='balanced')),
    ('Random Forest', RandomForestClassifier(random_state=42 ,class_weight='balanced')),
    ('CatBoost', CatBoostClassifier(random_state=42, verbose=0)),
    ('LightGBM', LGBMClassifier(random_state=42 ,class_weight='balanced',verbosity=-1))
]

for model in models:
    
    model_pipeline = Pipeline([
        ('Preprocessing', preprocessor),
        ('Model', model[1])
    ])

    result = cross_validate(
        model_pipeline,
        x,
        y,
        cv=cv,  
        scoring='balanced_accuracy',
        return_train_score=True
    )

    print(model[0])
    print(
        f"Train Balanced Accuracy: {result['train_score'].mean()*100:.2f}%"
    )
    print(
        f"Test Balanced Accuracy: "
        f"{result['test_score'].mean()*100:.2f}% "
        f"(± {result['test_score'].std()*100:.2f})"
    )
    print('-' * 50)
```

## 🔍 Hyperparameter Optimization

```python
# LGBMClassifier hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

para_grid = {
    'Model__n_estimators': randint(100, 1000),
    'Model__learning_rate': uniform(0.01, 0.29),   # 0.01 → 0.30
    'Model__max_depth': randint(3, 50),
    'Model__num_leaves': randint(20, 150)
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
model_pipline = Pipeline(steps=[('Preprocessing',preprocessor),('Model',LGBMClassifier(random_state=42,verbosity=-1,class_weight='balanced'))])
random_search = RandomizedSearchCV(estimator=model_pipline, param_distributions=para_grid, n_iter=10, cv=cv, scoring='balanced_accuracy', n_jobs=1, random_state=42)


random_search.fit(x, y)

print("Best Score:", random_search.best_score_)
print("Best Params:", random_search.best_params_)
```

## 🚀 Train Final Model

```python
final_pipeline = Pipeline(steps=[('Preprocessing',preprocessor),('Model',LGBMClassifier(n_estimators=152,num_leaves=87,max_depth=8,learning_rate=0.079472948293422))])
final_pipeline.fit(x, y) 
```

```python
# The following plot illustrates the twenty most influential features used by the LightGBM model.
feature_names = final_pipeline.named_steps[
    'Preprocessing'
].get_feature_names_out()

importance = final_pipeline.named_steps[
    'Model'
].feature_importances_

importance_df = (
    pd.DataFrame({
        'Feature': feature_names,
        'Importance': importance
    })
    .sort_values('Importance', ascending=False)
    .head(20)
)

fig = px.bar(
    importance_df,
    x='Importance',
    y='Feature',
    orientation='h',
    title='Top 20 Important Features'
)

fig.update_layout(yaxis={'categoryorder': 'total ascending'})

fig.show()
```
